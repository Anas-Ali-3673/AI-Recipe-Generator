```python
"""
Data Cleaning Module - Step 3
Merge NER and Extended_NER columns intelligently
"""

import ast
from typing import List, Set

def safe_parse_list(value: str) -> List[str]:
    """
    Safely parse string representation of list to actual list
    
    Args:
        value: String representation of list
        
    Returns:
        Parsed list or empty list if parsing fails
    """
    try:
        if pd.isna(value) or value == '':
            return []
        # Handle if already a list
        if isinstance(value, list):
            return value
        # Parse string representation
        parsed = ast.literal_eval(value)
        return parsed if isinstance(parsed, list) else []
    except:
        return []

def merge_ingredient_lists(ner: str, extended_ner: str) -> List[str]:
    """
    Merge NER and Extended_NER, keeping unique values (case-insensitive)
    
    Args:
        ner: NER column value
        extended_ner: Extended_NER column value
        
    Returns:
        Merged list of unique ingredients
    """
    # Parse both lists
    ner_list = safe_parse_list(ner)
    extended_ner_list = safe_parse_list(extended_ner)
    
    # Create case-insensitive unique set
    # Store in dict to maintain case of first occurrence
    unique_ingredients = {}
    
    for ingredient in ner_list + extended_ner_list:
        ingredient_clean = str(ingredient).strip()
        ingredient_lower = ingredient_clean.lower()
        
        # Only add if not empty and not already present (case-insensitive)
        if ingredient_clean and ingredient_lower not in unique_ingredients:
            unique_ingredients[ingredient_lower] = ingredient_clean
    
    # Return list of values (preserving first occurrence case)
    return list(unique_ingredients.values())

print("=" * 80)
print("🧹 DATA CLEANING - STEP 3: MERGE NER & EXTENDED_NER")
print("=" * 80)

# Test the function on first few rows
print("\n🔍 Testing merge function on sample rows:")
for idx in range(3):
    ner_val = df_cleaned.iloc[idx]['NER']
    extended_ner_val = df_cleaned.iloc[idx]['Extended_NER']
    merged = merge_ingredient_lists(ner_val, extended_ner_val)
    
    print(f"\n📝 Row {idx} - {df_cleaned.iloc[idx]['title'][:50]}:")
    print(f"   NER length: {len(safe_parse_list(ner_val))}")
    print(f"   Extended_NER length: {len(safe_parse_list(extended_ner_val))}")
    print(f"   Merged length: {len(merged)}")
    print(f"   Sample merged: {merged[:5]}...")

# Apply to entire dataset
print("\n⚙️ Merging NER and Extended_NER for all rows...")
print("   This may take a moment...")

df_cleaned['ingredients'] = df_cleaned.apply(
    lambda row: merge_ingredient_lists(row['NER'], row['Extended_NER']),
    axis=1
)

print("✅ Merge completed!")

# Statistics
print("\n📊 INGREDIENTS COLUMN STATISTICS:")
df_cleaned['ingredient_count'] = df_cleaned['ingredients'].apply(len)
print(f"   Mean ingredients per recipe: {df_cleaned['ingredient_count'].mean():.2f}")
print(f"   Median ingredients per recipe: {df_cleaned['ingredient_count'].median():.0f}")
print(f"   Min ingredients: {df_cleaned['ingredient_count'].min()}")
print(f"   Max ingredients: {df_cleaned['ingredient_count'].max()}")
print(f"   Recipes with 0 ingredients: {(df_cleaned['ingredient_count'] == 0).sum()}")

# Visualize distribution
fig = go.Figure()

fig.add_trace(go.Histogram(
    x=df_cleaned['ingredient_count'],
    nbinsx=50,
    marker_color='teal',
    name='Ingredient Count'
))

fig.update_layout(
    title="<b>Distribution of Ingredient Counts per Recipe</b>",
    xaxis_title="Number of Ingredients",
    yaxis_title="Frequency",
    height=500,
    showlegend=False
)

fig.show()

# Show sample of new column
print("\n👀 Sample of merged ingredients column:")
print(df_cleaned[['title', 'ingredients', 'ingredient_count']].head(3))
```

```python
"""
Data Cleaning Module - Step 6
Comprehensive Text Normalization for Recipe Dataset
Optimized for GPT-2 fine-tuning with proper text cleaning
"""

import re
import unicodedata
from typing import List
import gc

print("=" * 80)
print("🧹 DATA CLEANING - STEP 6: TEXT NORMALIZATION")
print("=" * 80)

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def normalize_text(text: str) -> str:
    """
    Comprehensive text normalization for recipe data
    
    Steps:
    1. Remove leading/trailing whitespace and tabs
    2. Normalize unicode characters
    3. Remove multiple spaces
    4. Fix common encoding issues
    5. Remove control characters
    
    Args:
        text: Raw text string
        
    Returns:
        Normalized text string
    """
    if pd.isna(text) or text == '':
        return ''
    
    # Convert to string if not already
    text = str(text)
    
    # Remove leading/trailing tabs and whitespace
    text = text.strip('\t\n\r ')
    
    # Normalize unicode (convert special characters to standard form)
    text = unicodedata.normalize('NFKC', text)
    
    # Fix common encoding issues
    text = text.replace('\u00a0', ' ')  # Non-breaking space
    text = text.replace('\u2009', ' ')  # Thin space
    text = text.replace('\u200b', '')   # Zero-width space
    text = text.replace('\ufeff', '')   # Zero-width no-break space
    
    # Remove control characters (except newline and tab)
    text = ''.join(char for char in text if unicodedata.category(char)[0] != 'C' or char in '\n\t')
    
    # Replace multiple spaces with single space
    text = re.sub(r' +', ' ', text)
    
    # Replace multiple newlines with single newline
    text = re.sub(r'\n+', '\n', text)
    
    # Remove leading/trailing whitespace again
    text = text.strip()
    
    return text


def normalize_ingredient_list(ingredients: List[str]) -> List[str]:
    """
    Normalize ingredient list by cleaning each ingredient
    
    Args:
        ingredients: List of ingredient strings
        
    Returns:
        Normalized list of ingredients
    """
    if not isinstance(ingredients, list):
        return []
    
    normalized = []
    for ingredient in ingredients:
        # Normalize each ingredient
        clean = normalize_text(str(ingredient))
        
        # Skip empty strings
        if clean:
            normalized.append(clean)
    
    return normalized


def normalize_directions(directions_str: str) -> List[str]:
    """
    Parse and normalize direction steps
    
    Args:
        directions_str: String representation of directions list
        
    Returns:
        List of normalized direction strings
    """
    try:
        # Parse string to list
        if isinstance(directions_str, list):
            directions = directions_str
        else:
            directions = ast.literal_eval(directions_str)
        
        if not isinstance(directions, list):
            return []
        
        # Normalize each direction step
        normalized = []
        for step in directions:
            clean = normalize_text(str(step))
            
            # Skip empty steps
            if clean:
                normalized.append(clean)
        
        return normalized
        
    except:
        return []


# ============================================================================
# APPLY TEXT NORMALIZATION
# ============================================================================

print("\n📝 Normalizing text fields...")
print("   This may take a few minutes for 2M+ recipes...\n")

# 1. NORMALIZE TITLE
print("   🔸 Normalizing titles...")
df_cleaned['title'] = df_cleaned['title'].apply(normalize_text)
print(f"      ✅ Completed: {len(df_cleaned)} titles normalized")

# Check for empty titles after normalization
empty_titles = df_cleaned['title'].str.len() == 0
if empty_titles.sum() > 0:
    print(f"      ⚠️  Found {empty_titles.sum()} empty titles after normalization")
    df_cleaned = df_cleaned[~empty_titles]
    print(f"      ✅ Removed empty titles. New count: {len(df_cleaned):,}")

# 2. NORMALIZE INGREDIENTS
print("\n   🔸 Normalizing ingredients...")
df_cleaned['ingredients'] = df_cleaned['ingredients'].apply(normalize_ingredient_list)
print(f"      ✅ Completed: {len(df_cleaned)} ingredient lists normalized")

# Recalculate ingredient count after normalization
df_cleaned['ingredient_count'] = df_cleaned['ingredients'].apply(len)
print(f"      ℹ️  Mean ingredients per recipe: {df_cleaned['ingredient_count'].mean():.2f}")

# 3. NORMALIZE DIRECTIONS
print("\n   🔸 Normalizing directions...")
df_cleaned['directions'] = df_cleaned['directions'].apply(normalize_directions)
df_cleaned['direction_count'] = df_cleaned['directions'].apply(len)
print(f"      ✅ Completed: {len(df_cleaned)} direction lists normalized")
print(f"      ℹ️  Mean directions per recipe: {df_cleaned['direction_count'].mean():.2f}")

# 4. NORMALIZE GENRE (already lowercase, but ensure clean)
print("\n   🔸 Normalizing genre...")
df_cleaned['genre'] = df_cleaned['genre'].apply(normalize_text)
print(f"      ✅ Completed: {df_cleaned['genre'].nunique()} unique genres")


# ============================================================================
# QUALITY CHECKS AFTER NORMALIZATION
# ============================================================================

print("\n" + "=" * 80)
print("🔍 QUALITY CHECKS AFTER NORMALIZATION")
print("=" * 80)

# Check for problematic recipes again
empty_ingredients = df_cleaned['ingredient_count'] == 0
empty_directions = df_cleaned['direction_count'] == 0
empty_titles = df_cleaned['title'].str.len() == 0

print(f"\n📊 Quality Metrics:")
print(f"   • Empty titles: {empty_titles.sum():,}")
print(f"   • Empty ingredients: {empty_ingredients.sum():,}")
print(f"   • Empty directions: {empty_directions.sum():,}")
print(f"   • Total problematic: {(empty_titles | empty_ingredients | empty_directions).sum():,}")

# Remove any problematic recipes
if (empty_titles | empty_ingredients | empty_directions).sum() > 0:
    initial_count = len(df_cleaned)
    df_cleaned = df_cleaned[~(empty_titles | empty_ingredients | empty_directions)]
    print(f"\n   🔧 Removed {initial_count - len(df_cleaned):,} problematic recipes")
    print(f"   ✅ Final count: {len(df_cleaned):,}")


# ============================================================================
# SAMPLE PREVIEW
# ============================================================================

print("\n" + "=" * 80)
print("👀 SAMPLE OF NORMALIZED DATA")
print("=" * 80)

for idx in range(3):
    print(f"\n📝 Recipe {idx + 1}:")
    print(f"   Title: {df_cleaned.iloc[idx]['title'][:60]}...")
    print(f"   Genre: {df_cleaned.iloc[idx]['genre']}")
    print(f"   Ingredients ({df_cleaned.iloc[idx]['ingredient_count']}): {df_cleaned.iloc[idx]['ingredients'][:3]}...")
    print(f"   Directions ({df_cleaned.iloc[idx]['direction_count']}): {df_cleaned.iloc[idx]['directions'][0][:80]}...")


# ============================================================================
# AGGRESSIVE MEMORY OPTIMIZATION
# ============================================================================

print("\n" + "=" * 80)
print("🧹 AGGRESSIVE MEMORY OPTIMIZATION")
print("=" * 80)

print("\n📊 BEFORE OPTIMIZATION:")
memory_before = df_cleaned.memory_usage(deep=True).sum() / 1024**2
print(f"   Total memory: {memory_before:.2f} MB")
for col in df_cleaned.columns:
    col_memory = df_cleaned[col].memory_usage(deep=True) / 1024**2
    print(f"   • {col}: {col_memory:.2f} MB")

# 1. Convert genre to category (saves significant memory)
print("\n🔸 Converting 'genre' to category dtype...")
df_cleaned['genre'] = df_cleaned['genre'].astype('category')
print("   ✅ Genre converted to category")

# 2. Optimize integer columns (use smallest possible dtype)
print("\n🔸 Optimizing integer columns...")
df_cleaned['ingredient_count'] = df_cleaned['ingredient_count'].astype('int16')
df_cleaned['direction_count'] = df_cleaned['direction_count'].astype('int16')
print("   ✅ Integer columns optimized to int16")

# 3. Clear any temporary columns or variables
print("\n🔸 Clearing temporary variables...")
temp_vars = ['empty_ingredients', 'empty_directions', 'empty_titles', 
             'initial_count', 'memory_before', 'col_memory', 'idx']
for var in temp_vars:
    if var in locals():
        del locals()[var]

# 4. Force garbage collection
gc.collect()
print("   ✅ Garbage collection completed")

# 5. Check memory after optimization
print("\n📊 AFTER OPTIMIZATION:")
memory_after = df_cleaned.memory_usage(deep=True).sum() / 1024**2
print(f"   Total memory: {memory_after:.2f} MB")
for col in df_cleaned.columns:
    col_memory = df_cleaned[col].memory_usage(deep=True) / 1024**2
    print(f"   • {col}: {col_memory:.2f} MB")

memory_saved = memory_before - memory_after
print(f"\n   💾 Memory saved: {memory_saved:.2f} MB ({memory_saved/memory_before*100:.1f}%)")


# ============================================================================
# FINAL SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("✅ TEXT NORMALIZATION & OPTIMIZATION COMPLETED!")
print("=" * 80)

print(f"\n📊 Final Dataset Statistics:")
print(f"   • Total recipes: {len(df_cleaned):,}")
print(f"   • Columns: {len(df_cleaned.columns)}")
print(f"   • Memory usage: {memory_after:.2f} MB")
print(f"   • Avg title length: {df_cleaned['title'].str.len().mean():.1f} chars")
print(f"   • Avg ingredients: {df_cleaned['ingredient_count'].mean():.1f}")
print(f"   • Avg directions: {df_cleaned['direction_count'].mean():.1f}")
print(f"   • Genres: {df_cleaned['genre'].nunique()}")

print(f"\n📋 Genre Distribution:")
print(df_cleaned['genre'].value_counts())

print("\n🎯 Dataset is now ready for GPT-2 fine-tuning!")
print("   Next step: Create training/validation splits")
```

```python
"""
GPT-2 Tokenization Module - FIXED FOR KAGGLE
Works offline with proper fallback mechanisms
Optimized for 10% dataset (~196k samples)
"""

import os
import pickle
import numpy as np
import torch
from torch.utils.data import Dataset
from tqdm.auto import tqdm
import gc
import warnings
warnings.filterwarnings('ignore')

print("=" * 80)
print("🔤 GPT-2 TOKENIZATION MODULE (OFFLINE-COMPATIBLE)")
print("=" * 80)

# ============================================================================
# CONFIGURATION
# ============================================================================

class TokenizationConfig:
    """Optimized configuration for tokenization"""
    
    # Model settings
    MODEL_NAME = "gpt2"
    MAX_LENGTH = 512
    
    # Special tokens
    BOS_TOKEN_ID = 50256
    EOS_TOKEN_ID = 50256
    PAD_TOKEN_ID = 50256
    
    # File paths
    SAVE_DIR = "/kaggle/working/tokenized_data1"
    TRAIN_FILE = "train_tokenized.npz"
    VAL_FILE = "val_tokenized.npz"
    TEST_FILE = "test_tokenized.npz"
    TOKENIZER_DIR = "tokenizer"
    
    # Processing settings
    BATCH_SIZE = 2000
    DTYPE = np.int32
    
    @classmethod
    def print_config(cls):
        print("\n⚙️ TOKENIZATION CONFIGURATION:")
        print(f"   Model: {cls.MODEL_NAME}")
        print(f"   Max sequence length: {cls.MAX_LENGTH} tokens")
        print(f"   Batch size: {cls.BATCH_SIZE}")
        print(f"   Data type: {cls.DTYPE.__name__}")
        print(f"   Storage: NumPy compressed arrays (.npz)")
        print(f"   Save directory: {cls.SAVE_DIR}")

config = TokenizationConfig()
config.print_config()

# Create save directory
os.makedirs(config.SAVE_DIR, exist_ok=True)

# ============================================================================
# MEMORY ESTIMATION
# ============================================================================

def estimate_memory_usage(num_samples, max_length, dtype=np.int32):
    """Estimate memory for tokenized data"""
    bytes_per_element = np.dtype(dtype).itemsize
    total_bytes = num_samples * max_length * bytes_per_element * 2
    total_mb = total_bytes / (1024**2)
    return total_mb

print("\n" + "=" * 80)
print("💾 MEMORY ESTIMATION")
print("=" * 80)

train_mem = estimate_memory_usage(len(train_df), config.MAX_LENGTH)
val_mem = estimate_memory_usage(len(val_df), config.MAX_LENGTH)
test_mem = estimate_memory_usage(len(test_df), config.MAX_LENGTH)
total_mem = train_mem + val_mem + test_mem

print(f"\nTokenized data storage:")
print(f"   Training:   {train_mem:>8.1f} MB ({len(train_df):,} samples)")
print(f"   Validation: {val_mem:>8.1f} MB ({len(val_df):,} samples)")
print(f"   Test:       {test_mem:>8.1f} MB ({len(test_df):,} samples)")
print(f"   {'='*45}")
print(f"   TOTAL:      {total_mem:>8.1f} MB (~{total_mem/1024:.2f} GB)")

print(f"\n💡 Memory during training:")
print(f"   Tokenized data:  ~{total_mem/1024:.2f} GB")
print(f"   Model (full):    ~3.5 GB")
print(f"   LoRA adapters:   ~0.5 GB")
print(f"   Training:        ~2.5 GB")
print(f"   {'='*45}")
estimated_peak = total_mem/1024 + 3.5 + 0.5 + 2.5 + 2
print(f"   Expected peak: ~{estimated_peak:.1f} GB / 19 GB")
print(f"   Safety margin: ~{19 - estimated_peak:.1f} GB")
print(f"\n✅ Memory usage is SAFE!")

# ============================================================================
# TOKENIZER SETUP - FIXED FOR OFFLINE USE
# ============================================================================

print("\n" + "=" * 80)
print("🔧 SETTING UP GPT-2 TOKENIZER (OFFLINE MODE)")
print("=" * 80)

def setup_tokenizer_offline():
    """
    Setup GPT-2 tokenizer with multiple fallback methods
    Priority: transformers (offline) > tiktoken
    """
    
    # Method 1: Try transformers with local_files_only
    try:
        from transformers import GPT2Tokenizer
        
        print("\n📥 Attempting offline tokenizer load...")
        
        # Try with local_files_only to avoid network errors
        tokenizer = GPT2Tokenizer.from_pretrained(
            'gpt2',
            local_files_only=True,  # Use cached files only
            add_prefix_space=False
        )
        
        tokenizer.pad_token = tokenizer.eos_token
        tokenizer.pad_token_id = tokenizer.eos_token_id
        
        print("✅ GPT2Tokenizer loaded (offline mode)")
        print(f"   Vocab size: {len(tokenizer):,}")
        return tokenizer, "transformers_offline"
        
    except Exception as e1:
        print(f"⚠️ Offline transformers failed: {str(e1)[:80]}")
        
        # Method 2: Try online with legacy mode (no chat templates)
        try:
            from transformers import GPT2Tokenizer
            
            print("\n🔄 Trying online mode (legacy)...")
            
            # Use trust_remote_code=False to avoid template issues
            tokenizer = GPT2Tokenizer.from_pretrained(
                'gpt2',
                trust_remote_code=False,
                add_prefix_space=False
            )
            
            tokenizer.pad_token = tokenizer.eos_token
            tokenizer.pad_token_id = tokenizer.eos_token_id
            
            print("✅ GPT2Tokenizer loaded (online mode)")
            print(f"   Vocab size: {len(tokenizer):,}")
            return tokenizer, "transformers_online"
            
        except Exception as e2:
            print(f"⚠️ Online transformers also failed: {str(e2)[:80]}")
            print("🔄 Falling back to tiktoken...")
    
    # Method 3: Tiktoken fallback (most reliable)
    try:
        import tiktoken
        
        print("\n📥 Loading tiktoken encoder...")
        
        class TiktokenWrapper:
            """Wrapper to make tiktoken compatible with transformers interface"""
            
            def __init__(self):
                self.encoding = tiktoken.get_encoding("gpt2")
                self.vocab_size = self.encoding.n_vocab
                self.eos_token_id = 50256
                self.bos_token_id = 50256
                self.pad_token_id = 50256
                self.eos_token = "<|endoftext|>"
                self.bos_token = "<|endoftext|>"
                self.pad_token = "<|endoftext|>"
            
            def encode(self, text, add_special_tokens=True):
                """Encode text to token IDs"""
                tokens = self.encoding.encode(text)
                if add_special_tokens:
                    tokens = [self.bos_token_id] + tokens + [self.eos_token_id]
                return tokens
            
            def __call__(self, texts, max_length=512, padding='max_length', 
                        truncation=True, return_tensors='pt', add_special_tokens=True):
                """Tokenize with transformers-like interface"""
                
                # Handle single text
                if isinstance(texts, str):
                    texts = [texts]
                
                all_input_ids = []
                all_attention_mask = []
                
                for text in texts:
                    # Encode
                    tokens = self.encoding.encode(text)
                    if add_special_tokens:
                        tokens = [self.bos_token_id] + tokens + [self.eos_token_id]
                    
                    # Truncate
                    if truncation and len(tokens) > max_length:
                        tokens = tokens[:max_length-1] + [self.eos_token_id]
                    
                    # Create attention mask
                    attention_mask = [1] * len(tokens)
                    
                    # Pad
                    if padding == 'max_length':
                        pad_length = max_length - len(tokens)
                        if pad_length > 0:
                            tokens = tokens + [self.pad_token_id] * pad_length
                            attention_mask = attention_mask + [0] * pad_length
                    
                    all_input_ids.append(tokens)
                    all_attention_mask.append(attention_mask)
                
                # Return format
                if return_tensors == 'pt':
                    return {
                        'input_ids': torch.tensor(all_input_ids, dtype=torch.long),
                        'attention_mask': torch.tensor(all_attention_mask, dtype=torch.long)
                    }
                else:
                    return {
                        'input_ids': all_input_ids,
                        'attention_mask': all_attention_mask
                    }
            
            def decode(self, tokens, skip_special_tokens=True):
                """Decode token IDs to text"""
                if isinstance(tokens, torch.Tensor):
                    tokens = tokens.tolist()
                
                # Handle batch
                if isinstance(tokens, list) and len(tokens) > 0 and isinstance(tokens[0], list):
                    return [self.decode(t, skip_special_tokens) for t in tokens]
                
                # Skip special tokens
                if skip_special_tokens:
                    tokens = [t for t in tokens if t not in [self.pad_token_id, self.eos_token_id, self.bos_token_id]]
                
                return self.encoding.decode(tokens)
            
            def __len__(self):
                return self.vocab_size
            
            def save_pretrained(self, path):
                """Save tokenizer config"""
                os.makedirs(path, exist_ok=True)
                config = {
                    'type': 'tiktoken',
                    'encoding': 'gpt2',
                    'vocab_size': self.vocab_size,
                    'special_tokens': {
                        'eos_token_id': self.eos_token_id,
                        'bos_token_id': self.bos_token_id,
                        'pad_token_id': self.pad_token_id
                    }
                }
                with open(os.path.join(path, 'tokenizer_config.json'), 'w') as f:
                    import json
                    json.dump(config, f, indent=2)
                print(f"   Tokenizer config saved to {path}")
        
        tokenizer = TiktokenWrapper()
        print("✅ Tiktoken wrapper initialized!")
        print(f"   Vocab size: {tokenizer.vocab_size:,}")
        return tokenizer, "tiktoken"
        
    except Exception as e3:
        raise RuntimeError(f"All tokenizer methods failed! Last error: {e3}")

# Load tokenizer
tokenizer, tokenizer_type = setup_tokenizer_offline()
print(f"\n✅ Using: {tokenizer_type}")

# Test tokenization
print("\n🧪 Testing tokenization...")
test_text = "Title: Chocolate Cake | Ingredients: flour, sugar, butter | Directions: Step 1: Mix."
test_result = tokenizer(test_text, max_length=64, padding='max_length', truncation=True, return_tensors='pt')

print(f"   Input: '{test_text[:50]}...'")
print(f"   Output shape: {test_result['input_ids'].shape}")
print(f"   Non-padding: {test_result['attention_mask'].sum().item()}")
print(f"   ✅ Tokenization working!")

# ============================================================================
# TOKENIZATION FUNCTION
# ============================================================================

def tokenize_dataset_optimized(df, tokenizer, max_length, batch_size, dtype, desc="Processing"):
    """
    Memory-efficient tokenization with robust error handling
    """
    print(f"\n🔄 Tokenizing {len(df):,} recipes...")
    print(f"   Batch size: {batch_size}")
    
    texts = df['text'].tolist()
    num_samples = len(texts)
    
    # Pre-allocate arrays
    input_ids = np.full((num_samples, max_length), config.PAD_TOKEN_ID, dtype=dtype)
    attention_mask = np.zeros((num_samples, max_length), dtype=dtype)
    
    successful = 0
    failed = 0
    
    # Process in batches
    for i in tqdm(range(0, len(texts), batch_size), desc=desc):
        batch_texts = texts[i:i + batch_size]
        batch_end = min(i + batch_size, num_samples)
        
        try:
            # Tokenize batch
            encoded = tokenizer(
                batch_texts,
                max_length=max_length,
                padding='max_length',
                truncation=True,
                return_tensors='pt',  # Always return PyTorch tensors
                add_special_tokens=True
            )
            
            # Convert to numpy
            batch_input_ids = encoded['input_ids'].numpy()
            batch_attention_mask = encoded['attention_mask'].numpy()
            
            # Store
            actual_size = batch_end - i
            input_ids[i:batch_end] = batch_input_ids[:actual_size]
            attention_mask[i:batch_end] = batch_attention_mask[:actual_size]
            
            successful += actual_size
            
        except Exception as e:
            # Fallback: process individually
            print(f"\n⚠️ Batch error: {str(e)[:80]}")
            for j, text in enumerate(batch_texts):
                idx = i + j
                try:
                    encoded = tokenizer(
                        text,
                        max_length=max_length,
                        padding='max_length',
                        truncation=True,
                        return_tensors='pt',
                        add_special_tokens=True
                    )
                    input_ids[idx] = encoded['input_ids'][0].numpy()
                    attention_mask[idx] = encoded['attention_mask'][0].numpy()
                    successful += 1
                except Exception as e2:
                    failed += 1
                    # Leave as PAD tokens
    
    print(f"\n   ✅ Successfully tokenized: {successful:,}/{num_samples:,}")
    if failed > 0:
        print(f"   ⚠️ Failed: {failed:,} (filled with PAD tokens)")
    
    # Memory usage
    memory_mb = (input_ids.nbytes + attention_mask.nbytes) / (1024**2)
    print(f"   💾 Memory used: {memory_mb:.1f} MB")
    
    # Labels = input_ids for causal LM
    labels = input_ids.copy()
    
    return {
        'input_ids': input_ids,
        'attention_mask': attention_mask,
        'labels': labels
    }

# ============================================================================
# TOKENIZE ALL DATASETS
# ============================================================================

print("\n" + "=" * 80)
print("💾 TOKENIZING DATASETS")
print("=" * 80)

def save_tokenized_npz(data_dict, filepath):
    """Save as compressed numpy file"""
    print(f"\n   💾 Saving to {filepath}...")
    np.savez_compressed(
        filepath,
        input_ids=data_dict['input_ids'],
        attention_mask=data_dict['attention_mask'],
        labels=data_dict['labels']
    )
    file_size = os.path.getsize(filepath) / (1024**2)
    print(f"   ✅ Saved ({file_size:.1f} MB)")

# Training set
print("\n" + "="*80)
print("📝 TRAINING SET")
print("="*80)
train_tokenized = tokenize_dataset_optimized(
    train_df, tokenizer, config.MAX_LENGTH, config.BATCH_SIZE, config.DTYPE, "Training"
)
train_path = os.path.join(config.SAVE_DIR, config.TRAIN_FILE)
save_tokenized_npz(train_tokenized, train_path)
del train_tokenized
gc.collect()

# Validation set
print("\n" + "="*80)
print("📝 VALIDATION SET")
print("="*80)
val_tokenized = tokenize_dataset_optimized(
    val_df, tokenizer, config.MAX_LENGTH, config.BATCH_SIZE, config.DTYPE, "Validation"
)
val_path = os.path.join(config.SAVE_DIR, config.VAL_FILE)
save_tokenized_npz(val_tokenized, val_path)
del val_tokenized
gc.collect()

# Test set
print("\n" + "="*80)
print("📝 TEST SET")
print("="*80)
test_tokenized = tokenize_dataset_optimized(
    test_df, tokenizer, config.MAX_LENGTH, config.BATCH_SIZE, config.DTYPE, "Test"
)
test_path = os.path.join(config.SAVE_DIR, config.TEST_FILE)
save_tokenized_npz(test_tokenized, test_path)

# Save test_df for evaluation
test_df.to_pickle(os.path.join(config.SAVE_DIR, 'test_df.pkl'))
print("   ✅ test_df.pkl saved for evaluation")

del test_tokenized
gc.collect()

# Save tokenizer
print(f"\n💾 Saving tokenizer...")
tokenizer_path = os.path.join(config.SAVE_DIR, config.TOKENIZER_DIR)
tokenizer.save_pretrained(tokenizer_path)
print(f"   ✅ Tokenizer saved to {tokenizer_path}")

# ============================================================================
# PYTORCH DATASET CLASS
# ============================================================================

class RecipeDataset(Dataset):
    """Memory-efficient PyTorch Dataset"""
    
    def __init__(self, tokenized_data_path: str, load_to_memory: bool = False):
        print(f"\n📂 Loading: {os.path.basename(tokenized_data_path)}")
        
        if load_to_memory:
            data = np.load(tokenized_data_path)
            self.input_ids = torch.from_numpy(data['input_ids']).long()
            self.attention_mask = torch.from_numpy(data['attention_mask']).long()
            self.labels = torch.from_numpy(data['labels']).long()
            mem_mb = (data['input_ids'].nbytes + data['attention_mask'].nbytes + data['labels'].nbytes) / (1024**2)
            print(f"   ✅ Loaded to RAM: {len(self.input_ids):,} samples ({mem_mb:.1f} MB)")
        else:
            self.data = np.load(tokenized_data_path, mmap_mode='r')
            self.input_ids = self.data['input_ids']
            self.attention_mask = self.data['attention_mask']
            self.labels = self.data['labels']
            print(f"   ✅ Memory-mapped: {len(self.input_ids):,} samples")
        
        self.load_to_memory = load_to_memory
    
    def __len__(self):
        return len(self.input_ids)
    
    def __getitem__(self, idx):
        if self.load_to_memory:
            return {
                'input_ids': self.input_ids[idx],
                'attention_mask': self.attention_mask[idx],
                'labels': self.labels[idx]
            }
        else:
            return {
                'input_ids': torch.from_numpy(np.array(self.input_ids[idx], dtype=np.int64)),
                'attention_mask': torch.from_numpy(np.array(self.attention_mask[idx], dtype=np.int64)),
                'labels': torch.from_numpy(np.array(self.labels[idx], dtype=np.int64))
            }

# ============================================================================
# VERIFICATION
# ============================================================================

print("\n" + "=" * 80)
print("🔍 VERIFICATION")
print("=" * 80)

# Load datasets
print("\n🔧 Loading datasets with memory mapping...")
train_dataset = RecipeDataset(train_path, load_to_memory=False)
val_dataset = RecipeDataset(val_path, load_to_memory=False)
test_dataset = RecipeDataset(test_path, load_to_memory=False)

print(f"\n✅ DATASETS READY:")
print(f"   Training:   {len(train_dataset):>8,} samples")
print(f"   Validation: {len(val_dataset):>8,} samples")
print(f"   Test:       {len(test_dataset):>8,} samples")

# Verify sample
sample = train_dataset[0]
print(f"\n👀 Sample verification:")
print(f"   Original: '{train_df['text'].iloc[0][:80]}...'")
print(f"   Shape: {sample['input_ids'].shape}")
print(f"   Dtype: {sample['input_ids'].dtype}")
print(f"   Non-padding: {sample['attention_mask'].sum().item()}/{config.MAX_LENGTH}")

decoded = tokenizer.decode(sample['input_ids'].tolist(), skip_special_tokens=True)
print(f"   Decoded: '{decoded[:80]}...'")
print(f"   ✅ Tokenization verified!")

# ============================================================================
# FINAL SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("🎉 TOKENIZATION COMPLETED!")
print("=" * 80)

total_size = sum([
    os.path.getsize(train_path),
    os.path.getsize(val_path),
    os.path.getsize(test_path)
]) / (1024**2)

print(f"\n📊 Summary:")
print(f"   Total recipes: {len(train_df) + len(val_df) + len(test_df):,}")
print(f"   Storage: {total_size:.1f} MB (compressed)")
print(f"   Max length: {config.MAX_LENGTH} tokens")
print(f"   Vocab size: {len(tokenizer):,}")
print(f"   Tokenizer: {tokenizer_type}")

print(f"\n📁 Saved files:")
print(f"   {train_path}")
print(f"   {val_path}")
print(f"   {test_path}")
print(f"   {tokenizer_path}")
print(f"   {os.path.join(config.SAVE_DIR, 'test_df.pkl')}")

print(f"\n🚀 Ready for training!")
print("=" * 80)
```

```python
"""
GPT-2 Full Precision LoRA Fine-tuning Module
FIXED: evaluation_strategy -> eval_strategy (transformers v4.30+)
Works with tiktoken tokenizer (offline-compatible)
Optimized for Kaggle 19GB RAM without bitsandbytes
Training time: ~6-8 hours for 1 epoch on 10% data (~196k recipes)
"""

import os
import gc
import json
import time
import torch
import numpy as np
from torch.utils.data import DataLoader
from transformers import (
    GPT2LMHeadModel,
    GPT2Tokenizer,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling,
    get_linear_schedule_with_warmup
)
from peft import (
    LoraConfig,
    get_peft_model,
    TaskType
)
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("=" * 80)
print("🚀 GPT-2 FULL PRECISION LORA FINE-TUNING (OFFLINE-COMPATIBLE)")
print("=" * 80)

# ============================================================================
# CONFIGURATION - OPTIMIZED FOR KAGGLE 19GB RAM
# ============================================================================

class TrainingConfig:
    """Complete training configuration for full precision training"""
    
    # Paths - UPDATED to match tokenization output
    TOKENIZED_DATA_DIR = "/kaggle/working/tokenized_data1"  # Match your tokenization dir
    OUTPUT_DIR = "/kaggle/working/gpt2_recipe_model"
    CHECKPOINT_DIR = os.path.join(OUTPUT_DIR, "checkpoints")
    LOGS_DIR = os.path.join(OUTPUT_DIR, "logs")
    FINAL_MODEL_DIR = os.path.join(OUTPUT_DIR, "final_model")
    
    # Model settings
    MODEL_NAME = "gpt2"
    MAX_LENGTH = 512
    
    # LoRA Configuration (Memory efficient - CRITICAL FOR FULL PRECISION)
    LORA_R = 8  # Rank (reduced from 16 to save memory)
    LORA_ALPHA = 16  # Scaling factor (typically 2*r)
    LORA_DROPOUT = 0.05
    LORA_TARGET_MODULES = ["c_attn"]  # Only attention (save memory)
    
    # Training Hyperparameters - OPTIMIZED FOR 11 HOURS
    NUM_EPOCHS = 1  # 1 epoch for safety (can increase if time permits)
    BATCH_SIZE = 1  # Per device (CRITICAL: must be 1 for memory)
    GRADIENT_ACCUMULATION_STEPS = 16  # Effective batch size = 1*16 = 16
    LEARNING_RATE = 2e-4
    WEIGHT_DECAY = 0.01
    WARMUP_RATIO = 0.03
    MAX_GRAD_NORM = 1.0
    
    # Memory optimization
    FP16 = True  # Mixed precision training (CRITICAL)
    OPTIM = "adamw_torch"  # Standard PyTorch optimizer
    GRADIENT_CHECKPOINTING = True  # Trade compute for memory (CRITICAL)
    
    # Checkpointing
    SAVE_STRATEGY = "steps"
    SAVE_STEPS = 3000  # Save every 3000 steps
    SAVE_TOTAL_LIMIT = 2  # Keep only 2 checkpoints (saves disk space)
    
    # Evaluation - FIXED: Use eval_strategy instead of evaluation_strategy
    EVAL_STRATEGY = "steps"  # Changed from evaluation_strategy
    EVAL_STEPS = 3000
    LOGGING_STEPS = 100
    
    # DataLoader
    DATALOADER_NUM_WORKERS = 2
    DATALOADER_PIN_MEMORY = True
    
    @classmethod
    def print_config(cls):
        print("\n⚙️ TRAINING CONFIGURATION (FULL PRECISION):")
        print(f"\n📦 Model:")
        print(f"   Base model: {cls.MODEL_NAME}")
        print(f"   Quantization: None (Full Precision)")
        print(f"   Mixed precision (FP16): {cls.FP16}")
        
        print(f"\n🔧 LoRA:")
        print(f"   Rank (r): {cls.LORA_R}")
        print(f"   Alpha: {cls.LORA_ALPHA}")
        print(f"   Dropout: {cls.LORA_DROPOUT}")
        print(f"   Target modules: {cls.LORA_TARGET_MODULES}")
        
        print(f"\n🎯 Training:")
        print(f"   Epochs: {cls.NUM_EPOCHS}")
        print(f"   Batch size: {cls.BATCH_SIZE}")
        print(f"   Gradient accumulation: {cls.GRADIENT_ACCUMULATION_STEPS}")
        print(f"   Effective batch size: {cls.BATCH_SIZE * cls.GRADIENT_ACCUMULATION_STEPS}")
        print(f"   Learning rate: {cls.LEARNING_RATE}")
        print(f"   Optimizer: {cls.OPTIM}")
        print(f"   FP16: {cls.FP16}")
        print(f"   Gradient checkpointing: {cls.GRADIENT_CHECKPOINTING}")
        
        print(f"\n💾 Checkpointing:")
        print(f"   Save every: {cls.SAVE_STEPS} steps")
        print(f"   Max checkpoints: {cls.SAVE_TOTAL_LIMIT}")
        print(f"   Output dir: {cls.OUTPUT_DIR}")

config = TrainingConfig()
config.print_config()

# Create directories
os.makedirs(config.OUTPUT_DIR, exist_ok=True)
os.makedirs(config.CHECKPOINT_DIR, exist_ok=True)
os.makedirs(config.LOGS_DIR, exist_ok=True)

# ============================================================================
# LOAD TOKENIZED DATASETS
# ============================================================================

print("\n" + "=" * 80)
print("📂 LOADING TOKENIZED DATASETS")
print("=" * 80)

from torch.utils.data import Dataset as TorchDataset

class RecipeDataset(TorchDataset):
    """Memory-efficient dataset with lazy loading"""
    
    def __init__(self, npz_path):
        print(f"   Loading: {os.path.basename(npz_path)}")
        self.data = np.load(npz_path, mmap_mode='r')  # Memory-mapped
        self.input_ids = self.data['input_ids']
        self.attention_mask = self.data['attention_mask']
        self.labels = self.data['labels']
        print(f"   ✅ Loaded: {len(self.input_ids):,} samples (memory-mapped)")
    
    def __len__(self):
        return len(self.input_ids)
    
    def __getitem__(self, idx):
        return {
            'input_ids': torch.from_numpy(self.input_ids[idx].astype(np.int64)),
            'attention_mask': torch.from_numpy(self.attention_mask[idx].astype(np.int64)),
            'labels': torch.from_numpy(self.labels[idx].astype(np.int64))
        }

# Load datasets
train_dataset = RecipeDataset(os.path.join(config.TOKENIZED_DATA_DIR, "train_tokenized.npz"))
val_dataset = RecipeDataset(os.path.join(config.TOKENIZED_DATA_DIR, "val_tokenized.npz"))

print(f"\n✅ Datasets loaded:")
print(f"   Training: {len(train_dataset):,} samples")
print(f"   Validation: {len(val_dataset):,} samples")

# ============================================================================
# LOAD TOKENIZER (OFFLINE-COMPATIBLE)
# ============================================================================

print("\n" + "=" * 80)
print("🔤 LOADING TOKENIZER (OFFLINE-COMPATIBLE)")
print("=" * 80)

def load_tokenizer_offline(tokenizer_dir):
    """Load tokenizer with multiple fallback methods"""
    
    # Check if tiktoken config exists
    config_path = os.path.join(tokenizer_dir, 'tokenizer_config.json')
    
    if os.path.exists(config_path):
        # Load tiktoken wrapper
        print("📥 Loading tiktoken tokenizer...")
        try:
            import tiktoken
            
            class TiktokenWrapper:
                """Tiktoken wrapper for training"""
                
                def __init__(self):
                    self.encoding = tiktoken.get_encoding("gpt2")
                    self.vocab_size = self.encoding.n_vocab
                    self.eos_token_id = 50256
                    self.bos_token_id = 50256
                    self.pad_token_id = 50256
                    self.eos_token = "<|endoftext|>"
                    self.bos_token = "<|endoftext|>"
                    self.pad_token = "<|endoftext|>"
                
                def encode(self, text, add_special_tokens=True):
                    tokens = self.encoding.encode(text)
                    if add_special_tokens:
                        tokens = [self.bos_token_id] + tokens + [self.eos_token_id]
                    return tokens
                
                def decode(self, tokens, skip_special_tokens=True):
                    if isinstance(tokens, torch.Tensor):
                        tokens = tokens.tolist()
                    if isinstance(tokens, list) and len(tokens) > 0 and isinstance(tokens[0], list):
                        return [self.decode(t, skip_special_tokens) for t in tokens]
                    if skip_special_tokens:
                        tokens = [t for t in tokens if t not in [self.pad_token_id, self.eos_token_id, self.bos_token_id]]
                    return self.encoding.decode(tokens)
                
                def __len__(self):
                    return self.vocab_size
                
                def __call__(self, *args, **kwargs):
                    # For compatibility with DataCollator
                    raise NotImplementedError("Use tokenized data directly")
            
            tokenizer = TiktokenWrapper()
            print("✅ Tiktoken tokenizer loaded!")
            return tokenizer, "tiktoken"
            
        except Exception as e:
            print(f"⚠️ Tiktoken failed: {e}")
    
    # Fallback: Try transformers (offline first)
    try:
        print("📥 Trying transformers tokenizer (offline)...")
        tokenizer = GPT2Tokenizer.from_pretrained(
            'gpt2',
            local_files_only=True
        )
        tokenizer.pad_token = tokenizer.eos_token
        tokenizer.pad_token_id = tokenizer.eos_token_id
        print("✅ Transformers tokenizer loaded (offline)")
        return tokenizer, "transformers"
    except:
        pass
    
    # Last resort: online
    try:
        print("📥 Trying transformers tokenizer (online)...")
        tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
        tokenizer.pad_token = tokenizer.eos_token
        tokenizer.pad_token_id = tokenizer.eos_token_id
        print("✅ Transformers tokenizer loaded (online)")
        return tokenizer, "transformers"
    except Exception as e:
        raise RuntimeError(f"Failed to load any tokenizer: {e}")

tokenizer, tokenizer_type = load_tokenizer_offline(
    os.path.join(config.TOKENIZED_DATA_DIR, "tokenizer")
)
print(f"   Vocab size: {len(tokenizer):,}")
print(f"   Type: {tokenizer_type}")

# ============================================================================
# LOAD MODEL IN FULL PRECISION
# ============================================================================

print("\n" + "=" * 80)
print("🤖 LOADING GPT-2 MODEL (FULL PRECISION)")
print("=" * 80)

print("📥 Loading GPT-2 model...")
print(f"   Mode: Full precision (no quantization)")
print(f"   Mixed precision: FP16 enabled")

# Load model - try offline first, then online
try:
    print("   Attempting offline load...")
    model = GPT2LMHeadModel.from_pretrained(
        config.MODEL_NAME,
        local_files_only=True,
        torch_dtype=torch.float32
    )
    print("   ✅ Model loaded (offline)")
except:
    print("   Offline failed, trying online...")
    model = GPT2LMHeadModel.from_pretrained(
        config.MODEL_NAME,
        torch_dtype=torch.float32
    )
    print("   ✅ Model loaded (online)")

# Move to GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"   Device: {device}")

if device == "cuda":
    model = model.to(device)
    print(f"   GPU: {torch.cuda.get_device_name(0)}")

# Enable gradient checkpointing (CRITICAL for memory)
if config.GRADIENT_CHECKPOINTING:
    model.gradient_checkpointing_enable()
    print("✅ Gradient checkpointing enabled")

# ============================================================================
# APPLY LORA
# ============================================================================

print("\n" + "=" * 80)
print("🔗 APPLYING LORA ADAPTERS")
print("=" * 80)

lora_config = LoraConfig(
    r=config.LORA_R,
    lora_alpha=config.LORA_ALPHA,
    target_modules=config.LORA_TARGET_MODULES,
    lora_dropout=config.LORA_DROPOUT,
    bias="none",
    task_type=TaskType.CAUSAL_LM
)

print("🔧 LoRA Configuration:")
print(f"   Rank: {config.LORA_R}")
print(f"   Alpha: {config.LORA_ALPHA}")
print(f"   Target modules: {config.LORA_TARGET_MODULES}")

model = get_peft_model(model, lora_config)
print("✅ LoRA adapters applied!")

# Print trainable parameters
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
total_params = sum(p.numel() for p in model.parameters())
print(f"\n📊 Model Parameters:")
print(f"   Trainable: {trainable_params:,} ({trainable_params/total_params*100:.2f}%)")
print(f"   Total: {total_params:,}")
print(f"   Estimated memory: ~3.5 GB (full precision) + ~0.5 GB (LoRA)")

# ============================================================================
# DATA COLLATOR - SIMPLE VERSION (NO TOKENIZER NEEDED)
# ============================================================================

class SimpleDataCollator:
    """
    Simple data collator that doesn't require tokenizer
    Works with pre-tokenized data
    """
    def __call__(self, features):
        batch = {
            'input_ids': torch.stack([f['input_ids'] for f in features]),
            'attention_mask': torch.stack([f['attention_mask'] for f in features]),
            'labels': torch.stack([f['labels'] for f in features])
        }
        return batch

data_collator = SimpleDataCollator()
print("\n✅ Data collator initialized (no tokenizer required)")

# ============================================================================
# TRAINING ARGUMENTS - FIXED VERSION
# ============================================================================

print("\n" + "=" * 80)
print("⚙️ SETTING UP TRAINING ARGUMENTS")
print("=" * 80)

training_args = TrainingArguments(
    # Output
    output_dir=config.OUTPUT_DIR,
    
    # Training
    num_train_epochs=config.NUM_EPOCHS,
    per_device_train_batch_size=config.BATCH_SIZE,
    per_device_eval_batch_size=config.BATCH_SIZE,
    gradient_accumulation_steps=config.GRADIENT_ACCUMULATION_STEPS,
    
    # Optimization
    learning_rate=config.LEARNING_RATE,
    weight_decay=config.WEIGHT_DECAY,
    warmup_ratio=config.WARMUP_RATIO,
    max_grad_norm=config.MAX_GRAD_NORM,
    optim=config.OPTIM,
    
    # Mixed precision
    fp16=config.FP16,
    fp16_full_eval=False,
    
    # Logging
    logging_dir=config.LOGS_DIR,
    logging_steps=config.LOGGING_STEPS,
    logging_first_step=True,
    
    # Evaluation - FIXED: Only use eval_strategy (not evaluation_strategy)
    eval_strategy=config.EVAL_STRATEGY,  # CHANGED: Removed evaluation_strategy
    eval_steps=config.EVAL_STEPS,
    
    # Checkpointing
    save_strategy=config.SAVE_STRATEGY,
    save_steps=config.SAVE_STEPS,
    save_total_limit=config.SAVE_TOTAL_LIMIT,
    
    # DataLoader
    dataloader_num_workers=config.DATALOADER_NUM_WORKERS,
    dataloader_pin_memory=config.DATALOADER_PIN_MEMORY,
    
    # Misc
    report_to="none",  # Disable wandb/tensorboard
    remove_unused_columns=False,
    load_best_model_at_end=False,
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    
    # Memory optimization
    gradient_checkpointing=config.GRADIENT_CHECKPOINTING,
    ddp_find_unused_parameters=False,
)

print("✅ Training arguments configured")

# Calculate training stats
total_steps = (len(train_dataset) // (config.BATCH_SIZE * config.GRADIENT_ACCUMULATION_STEPS)) * config.NUM_EPOCHS
print(f"\n📊 Training Statistics:")
print(f"   Total steps: {total_steps:,}")
print(f"   Steps per epoch: {total_steps // config.NUM_EPOCHS:,}")
print(f"   Warmup steps: {int(total_steps * config.WARMUP_RATIO):,}")
print(f"   Save every: {config.SAVE_STEPS:,} steps")
print(f"   Eval every: {config.EVAL_STEPS:,} steps")

# Estimate time
estimated_time_hours = (total_steps * 0.4) / 3600  # ~0.4 sec per step (conservative)
print(f"   Estimated time: ~{estimated_time_hours:.1f} hours")

# ============================================================================
# TRAINER
# ============================================================================

print("\n" + "=" * 80)
print("🏋️ INITIALIZING TRAINER")
print("=" * 80)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    data_collator=data_collator,
)

print("✅ Trainer initialized")

# ============================================================================
# MEMORY CHECK BEFORE TRAINING
# ============================================================================

print("\n" + "=" * 80)
print("💾 MEMORY CHECK")
print("=" * 80)

if torch.cuda.is_available():
    gpu_memory = torch.cuda.get_device_properties(0).total_memory / 1024**3
    print(f"   GPU memory: {gpu_memory:.1f} GB")
    torch.cuda.empty_cache()
    gc.collect()
    print("   ✅ GPU cache cleared")

# Estimate peak memory
print(f"\n📊 Estimated peak RAM usage:")
print(f"   Tokenized data (mmap): ~0.1 GB")
print(f"   Model (full precision): ~3.5 GB")
print(f"   LoRA adapters: ~0.5 GB")
print(f"   Optimizer states: ~2.5 GB")
print(f"   Training overhead: ~2.0 GB")
print(f"   Batch in memory: ~0.3 GB")
print(f"   {'='*45}")
print(f"   TOTAL: ~8.9 GB / 19 GB")
print(f"   Safety margin: ~10 GB")
print(f"\n✅ SAFE to proceed!")

# ============================================================================
# START TRAINING
# ============================================================================

print("\n" + "=" * 80)
print("🚀 STARTING TRAINING")
print("=" * 80)
print(f"\n⏰ Start time: {time.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"⏱️ Estimated duration: ~{estimated_time_hours:.1f} hours")
print(f"📊 Training on {len(train_dataset):,} recipes")
print(f"🎯 {config.NUM_EPOCHS} epoch(s) × {total_steps // config.NUM_EPOCHS:,} steps/epoch")
print("\n" + "=" * 80 + "\n")

# Start training
start_time = time.time()

try:
    train_result = trainer.train()
    
    training_time = time.time() - start_time
    
    print("\n" + "=" * 80)
    print("✅ TRAINING COMPLETED SUCCESSFULLY!")
    print("=" * 80)
    print(f"\n⏱️ Training time: {training_time/3600:.2f} hours")
    print(f"📊 Final training loss: {train_result.training_loss:.4f}")
    print(f"🎯 Total steps: {train_result.global_step:,}")
    
    # Save training metrics
    metrics = {
        "training_loss": float(train_result.training_loss),
        "training_time_hours": training_time / 3600,
        "total_steps": train_result.global_step,
        "samples_processed": len(train_dataset) * config.NUM_EPOCHS
    }
    
    with open(os.path.join(config.OUTPUT_DIR, "training_metrics.json"), "w") as f:
        json.dump(metrics, f, indent=2)
    
    print(f"\n💾 Metrics saved to {config.OUTPUT_DIR}/training_metrics.json")

except Exception as e:
    print(f"\n❌ Training failed: {str(e)}")
    import traceback
    traceback.print_exc()
    raise

# ============================================================================
# SAVE FINAL MODEL
# ============================================================================

print("\n" + "=" * 80)
print("💾 SAVING FINAL MODEL")
print("=" * 80)

# Save LoRA adapters
print("\n🔧 Saving LoRA adapters...")
model.save_pretrained(config.FINAL_MODEL_DIR)
print(f"   ✅ LoRA adapters saved to {config.FINAL_MODEL_DIR}")

# Save tokenizer (if transformers tokenizer)
if tokenizer_type == "transformers":
    tokenizer.save_pretrained(config.FINAL_MODEL_DIR)
    print(f"   ✅ Tokenizer saved")
else:
    # Save tiktoken config
    import json
    tokenizer_config = {
        'type': 'tiktoken',
        'encoding': 'gpt2',
        'vocab_size': len(tokenizer)
    }
    with open(os.path.join(config.FINAL_MODEL_DIR, 'tokenizer_config.json'), 'w') as f:
        json.dump(tokenizer_config, f, indent=2)
    print(f"   ✅ Tokenizer config saved")

# Save training config
training_config_dict = {
    "model_name": config.MODEL_NAME,
    "max_length": config.MAX_LENGTH,
    "lora_r": config.LORA_R,
    "lora_alpha": config.LORA_ALPHA,
    "num_epochs": config.NUM_EPOCHS,
    "batch_size": config.BATCH_SIZE,
    "learning_rate": config.LEARNING_RATE,
    "training_samples": len(train_dataset),
    "training_time_hours": training_time / 3600,
    "precision": "full (fp32) + fp16 mixed precision"
}

with open(os.path.join(config.FINAL_MODEL_DIR, "training_config.json"), "w") as f:
    json.dump(training_config_dict, f, indent=2)

print(f"   ✅ Training config saved")

# ============================================================================
# FINAL EVALUATION
# ============================================================================

print("\n" + "=" * 80)
print("📊 FINAL EVALUATION")
print("=" * 80)

print("\n🔍 Evaluating on validation set...")
eval_results = trainer.evaluate()

print(f"\n📊 Validation Results:")
print(f"   Loss: {eval_results['eval_loss']:.4f}")
print(f"   Perplexity: {np.exp(eval_results['eval_loss']):.2f}")

# Save evaluation results
eval_results_serializable = {k: float(v) if isinstance(v, (np.floating, np.integer)) else v 
                             for k, v in eval_results.items()}
with open(os.path.join(config.OUTPUT_DIR, "eval_results.json"), "w") as f:
    json.dump(eval_results_serializable, f, indent=2)

print(f"\n💾 Results saved to {config.OUTPUT_DIR}/eval_results.json")

# ============================================================================
# GENERATE SAMPLE RECIPES (QUICK TEST)
# ============================================================================

print("\n" + "=" * 80)
print("🧪 GENERATING SAMPLE RECIPES")
print("=" * 80)

# Set model to eval mode
model.eval()

# Test prompts
test_prompts = [
    "Title: Chocolate Chip Cookies | Ingredients:",
    "Title: Chicken Curry | Ingredients:",
    "Title: Vegetable Soup | Ingredients:"
]

print("\n🎨 Generating samples (this may take a moment)...\n")

for i, prompt in enumerate(test_prompts):
    print(f"{'='*80}")
    print(f"Sample {i+1}:")
    print(f"{'='*80}")
    print(f"Prompt: {prompt}\n")
    
    # Tokenize
    input_ids = torch.tensor([tokenizer.encode(prompt)]).to(device)
    
    # Generate
    with torch.no_grad():
        output = model.generate(
            input_ids,
            max_length=200,
            num_return_sequences=1,
            temperature=0.8,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.pad_token_id
        )
    
    # Decode
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    print(f"Generated:\n{generated_text}\n")

# ============================================================================
# CLEANUP & SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("🧹 CLEANUP & SUMMARY")
print("=" * 80)

# Clear cache
if torch.cuda.is_available():
    torch.cuda.empty_cache()
gc.collect()
print("✅ Memory cleared")

# Print summary
print(f"\n📊 FINAL SUMMARY:")
print(f"{'='*80}")
print(f"✅ Training completed successfully!")
print(f"⏱️ Total time: {training_time/3600:.2f} hours")
print(f"📦 Model saved to: {config.FINAL_MODEL_DIR}")
print(f"💾 Checkpoints: {config.CHECKPOINT_DIR}")
print(f"📈 Logs: {config.LOGS_DIR}")
print(f"📊 Validation loss: {eval_results['eval_loss']:.4f}")
print(f"📊 Perplexity: {np.exp(eval_results['eval_loss']):.2f}")

# Check disk usage
import subprocess
try:
    result = subprocess.run(['du', '-sh', config.OUTPUT_DIR], 
                          capture_output=True, text=True)
    disk_usage = result.stdout.split()[0]
    print(f"💾 Total disk usage: {disk_usage}")
except:
    pass

print(f"\n🎉 ALL DONE! Model is ready for inference.")
print(f"{'='*80}")

# Save final status
final_status = {
    "status": "completed",
    "training_time_hours": training_time / 3600,
    "final_loss": float(eval_results['eval_loss']),
    "perplexity": float(np.exp(eval_results['eval_loss'])),
    "model_path": config.FINAL_MODEL_DIR,
    "timestamp": time.strftime('%Y-%m-%d %H:%M:%S'),
    "precision": "full (fp32) + fp16 mixed precision",
    "tokenizer_type": tokenizer_type
}

with open(os.path.join(config.OUTPUT_DIR, "final_status.json"), "w") as f:
    json.dump(final_status, f, indent=2)

print(f"\n✅ Status saved to {config.OUTPUT_DIR}/final_status.json")
```
